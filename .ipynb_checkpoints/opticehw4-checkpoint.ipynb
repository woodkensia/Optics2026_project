```python
import matplotlib.pyplot as plt
import numpy as np
from astropy.io import fits

file = '110906925_filtered.fits'

with fits.open(file) as hdul:
    
    photon_data = hdul[1].data
    
    timestamps = photon_data['TIME']


time_norm = timestamps - timestamps.min()
 
# Calculate how many total bins we need to cover the whole observation
total_duration = time_norm.max() - time_norm.min()
bin_size_seconds= 50
number_of_bins = int(total_duration / bin_size_seconds)


plt.figure(figsize=(10, 6))
counts, bins, patches = plt.hist(time_norm, bins= number_of_bins, color='#F4B33B', edgecolor='black', alpha=0.7)

# 5. Make it look professional
plt.xlabel(f'Time since observation start (s)')
plt.ylabel(f'Photon Counts per {bin_size_seconds}s bin')
plt.title('Fermi-LAT Solar Flare Light Curve')
plt.savefig('fermidataQ2.png')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from astropy.io import fits

file = '110906925_filtered.fits'

with fits.open(file) as hdul:
    
    photon_data = hdul[1].data
    
    timestamps = photon_data['TIME']


time_norm = timestamps - timestamps.min()
 
# Calculate how many total bins we need to cover the whole observation
total_duration = time_norm.max() - time_norm.min()
bin_size_seconds= 100
number_of_bins = int(total_duration / bin_size_seconds)


plt.figure(figsize=(10, 6))
counts, bins, patches = plt.hist(time_norm, bins= number_of_bins, color='#F4B33B', edgecolor='black', alpha=0.7)

# 5. Make it look professional
plt.xlabel(f'Time since observation start (s)')
plt.ylabel(f'Photon Counts per {bin_size_seconds}s bin')
plt.title('Fermi-LAT Solar Flare Light Curve')
plt.savefig('bin100.png')
plt.show()
```

```python
print("This is the mean", np.mean(counts))
print("This is the varirance", np.var(counts))
print("This is the Standard Deviation", np.std(counts))
```

```python
peak_counts = np.max(counts)
peak_error = np.sqrt(peak_counts)
print( "This is the Max photons in a single bin", peak_counts)
print( "This is the statistical error of the peak", peak_error)
```
